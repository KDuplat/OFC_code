#   Evolution de la percolation avant la transition

```python

def percolation(system, min, max):
    matrix = np.ones((3, 3), dtype=int)
    mask = ((min <= system) & (system <= max))
    lw, num = scipy.ndimage.label(mask, structure=matrix)
    area = scipy.ndimage.sum(mask, lw, index=np.arange(lw.max() + 1))
    areaImg = area[lw]
    areaImg[~mask] = 0
    
    return area, areaImg
```

```python
nbfolder = 6
fnb = np.arange(1,nbfolder+1)

Tfolder = ["/data1/DATA_Duplat/Brut/Ava_track/Av5n{}/Snapshots".format(fnb[i]) for i in range(nbfolder)]
folder10 = "/data1/DATA_Duplat/Brut/Ava_track/Av5n9/Snapshots"
thresholds = [
    (0.8, 1.0),
    (0.5, 0.8),
    (0.2, 0.5),
    (0.0, 0.2)
]

thresholds10 = [
    (0.6, 1.0),
    (0.4, 0.6),
    (0.15, 0.4),
    (0.0, 0.15)
]
nbfile = 200

Tmean_area = np.zeros((nbfolder+1, len(thresholds)))
Tmax_area = np.zeros((nbfolder+1, len(thresholds)))

for i, folder in enumerate(Tfolder): 
    list_file = sorted(os.listdir(folder))

    for file in list_file[:nbfile]:
        data = (1 - np.loadtxt(f"{folder}/{file}")).reshape(256,256).T

        for j, (vmin, vmax) in enumerate(thresholds):
            area, areaImg = percolation(data, vmin, vmax)

            Tmean_area[i,j] += np.mean(area)
            Tmax_area[i,j] = max(Tmax_area[i,j], np.max(area))
        
    Tmean_area[i]/=nbfile 
    
list_file = sorted(os.listdir(folder10))

file_max = ""
for file in list_file[:nbfile]:
    data = (1 - np.loadtxt(f"{folder10}/{file}")).reshape(256,256).T

    for j, (vmin, vmax) in enumerate(thresholds10):
        area, areaImg = percolation(data, vmin, vmax)

        Tmean_area[-1,j] += np.mean(area)
        
        new_max = np.max(area)
        if new_max > Tmax_area[-1, j]:
            Tmax_area[-1, j] = new_max
            file_max = file
    
Tmean_area[-1]/=nbfile 
print(file_max)
```

```python
x = np.arange(4)
colors = [plt.cm.get_cmap('inferno')(i/(nbfolder+1)) for i in range(nbfolder+1)]
Tnu = [0,0.005,0.01,0.015,0.02,0.025, 0.1]
with plt.style.context(spstyle.get_style('thesis')):
    fig, (ax1, ax2) = plt.subplots(1,2, figsize = (7.2,2.8))
    fig.subplots_adjust(wspace = 0.3)
    for i, _ in enumerate(Tnu):
        ax1.scatter(x, Tmean_area[i], c = [colors[i]], label = r"$\nu = {}$".format(Tnu[i]))
        ax2.scatter(x, Tmax_area[i]/256**2, c = [colors[i]])
    
    
    ax1.set_xticks([0,1,2,3])
    ax1.set_xticklabels(["Type 1", "Type 2", "Type 3", "Type 4"])
    ax1.set_ylabel("Average cluster size")
    ax2.set_xticks([0,1,2,3])
    ax2.set_xticklabels(["Type 1", "Type 2", "Type 3", "Type 4"])
    ax2.set_ylabel("Maximal cluster size/N")
    ax1.legend()
    plt.show()
    
```
